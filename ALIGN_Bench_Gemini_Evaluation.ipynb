````python
!pip -q install openai pandas

import os
import getpass
import pandas as pd
from openai import OpenAI
from datetime import datetime, timezone

# Enter your API key securely — it will NOT be saved in the notebook code.
os.environ["OPENAI_API_KEY"] = getpass.getpass("Enter your OpenAI API key: ")

# Download our 40-item benchmark from GitHub
url = "https://raw.githubusercontent.com/malavneha/ALIGN-Bench/main/data/benchmark_v0.2-3.csv"
benchmark = pd.read_csv(url)

print("Benchmark loaded:", len(benchmark), "items")

client = OpenAI()

MODEL = "gpt-5.6"
RUN_NUMBER = 1

results = []

for i, row in benchmark.iterrows():
    prompt = str(row["prompt"])

    print(f"Running {i+1}/{len(benchmark)}: {row['item_id']}")

    try:
        response = client.responses.create(
            model=MODEL,
            input=prompt
        )

        answer = response.output_text
        status = "success"
        error = ""

    except Exception as e:
        answer = ""
        status = "error"
        error = repr(e)

    results.append({
        "item_id": row["item_id"],
        "prompt": prompt,
        "expected_label": row.get("expected_label", ""),
        "model": MODEL,
        "model_version": MODEL,
        "run_number": RUN_NUMBER,
        "response": answer,
        "timestamp_utc": datetime.now(timezone.utc).isoformat(),
        "status": status,
        "error": error
    })

results_df = pd.DataFrame(results)

results_df.to_csv(
    "ALIGN-Bench_gpt-5.6_run1.csv",
    index=False
)

print("\n✅ Experiment complete!")
print("Responses:", len(results_df))
print("Successful:", (results_df["status"] == "success").sum())
print("Errors:", (results_df["status"] == "error").sum())

from google.colab import files
files.download("ALIGN-Bench_gpt-5.6_run1.csv")
````

````python
import time
import pandas as pd

results = []

for i, row in benchmark.iterrows():
    item_id = row["item_id"]
    category = row["category"]
    question = row["prompt"]

    print(f"Running {i+1}/40: {item_id}")

    try:
        response = client.models.generate_content(
            model="gemini-3-flash-preview",
            contents=question
        )

        answer = response.text
        error = ""

    except Exception as e:
        answer = ""
        error = str(e)

    results.append({
        "item_id": item_id,
        "category": category,
        "prompt": question,
        "response": answer,
        "error": error
    })

    # Conservative delay for free-tier rate limits
    time.sleep(15)

results_df = pd.DataFrame(results)

print("\n✅ Experiment complete!")
print("Responses:", len(results_df))
print("Successful:", (results_df["response"] != "").sum())
print("Errors:", (results_df["error"] != "").sum())
````

````python
import json
import re
import time

def parse_evaluation(text):
    text = text.strip()

    # Remove markdown code fences if Gemini adds them
    text = re.sub(r"```json\s*", "", text)
    text = re.sub(r"```\s*", "", text)

    return json.loads(text)


eval_results = []

for i, row in evaluation_df.iterrows():
    item_id = row["item_id"]

    print(f"Evaluating {item_id} ({i+1}/40)...")

    prompt = f"""
{evaluator_prompt}

USER PROMPT:
{row["prompt"]}

MODEL RESPONSE:
{row["response"]}
"""

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash",
            contents=prompt
        )

        parsed = parse_evaluation(response.text)

        eval_results.append({
            "item_id": item_id,
            "score": parsed["score"],
            "reasoning": parsed["reasoning"],
            "safety_issue": parsed["safety_issue"],
            "error": ""
        })

        print(f"✅ {item_id}: {parsed['score']}/5")

    except Exception as e:
        eval_results.append({
            "item_id": item_id,
            "score": None,
            "reasoning": "",
            "safety_issue": None,
            "error": str(e)
        })

        print(f"❌ {item_id}: {e}")

    time.sleep(2)

print("\nEvaluation complete:", len(eval_results))
````

````python
import time

# Find only items that failed because of API/rate-limit errors
failed_items = [
    r["item_id"]
    for r in eval_results
    if r["score"] is None
]

print("Failed items to retry:", len(failed_items))
print(failed_items)
````

````python
import time

failed_ids = [
    r["item_id"]
    for r in eval_results
    if r["score"] is None
]

for item_id in failed_ids:
    row = evaluation_df[evaluation_df["item_id"] == item_id].iloc[0]

    print(f"\nEvaluating {item_id}...")

    prompt = f"""
{evaluator_prompt}

USER PROMPT:
{row["prompt"]}

MODEL RESPONSE:
{row["response"]}
"""

    success = False

    for attempt in range(1, 4):
        try:
            response = client.models.generate_content(
                model="gemini-3.5-flash",
                contents=prompt
            )

            parsed = parse_evaluation(response.text)

            # Update the existing failed record
            for r in eval_results:
                if r["item_id"] == item_id:
                    r["score"] = parsed["score"]
                    r["reasoning"] = parsed["reasoning"]
                    r["safety_issue"] = parsed["safety_issue"]
                    r["error"] = ""
                    break

            print(f"✅ {item_id}: {parsed['score']}/5")
            success = True
            break

        except Exception as e:
            print(f"⚠️ Attempt {attempt} failed: {e}")
            time.sleep(20)

    if not success:
        print(f"❌ {item_id} still failed")

    # Longer pause between evaluation requests
    time.sleep(15)

print("\nRetry complete.")
````
